A pós indentificar todos os produtos unicos crie uma separção para poder inserir os resultados na tabela Generico aonde diversos produtos não foram classificados

```python
palavras_genericos = [
    'CLORIDRATO', 'SÓDICO', 'MALEATO', 'BESILATO', 'ACETATO', 'FUMARATO',
    'NITRATO', 'BROMETO', 'TARTARATO', 'OLAMINA', 'CÁLCIO', 'MAGNÉSIO',
    'ÁCIDO', 'SULFATO', 'SÓDIO', 'POTÁSSICO', 'CLORETO'
]
```

```python
def eh_generico(nome):
    nome = nome.upper()
    
    # se contém palavras típicas de princípio ativo → genérico
    for p in palavras_genericos:
        if p in nome:
            return 'Sim'
    
    # se contém palavras que indicam material/equipamento → não é medicamento
    materiais = ['SERINGA', 'CATETER', 'LUVA', 'TUBO', 'APARELHO', 'MÁSCARA',
                 'EQUIPO', 'FRASCO', 'BROCA', 'AGULHA', 'CURATIVO', 'VENTILADOR',
                 'CÂNULA', 'DRENO', 'BOLSA', 'COMPRESSA', 'ODONTOLÓGICO']
    
    for m in materiais:
        if m in nome:
            return 'Não é medicamento'
    
    # se não caiu em nenhuma regra, assume não genérico
    return 'Não'

df['generico'] = df['nome_produto'].apply(eh_generico) 
# apply() aplica a função eh_generico() em cada elemento da coluna 'nome_produto' e cria uma nova coluna 'generico' com os resultados.
```
