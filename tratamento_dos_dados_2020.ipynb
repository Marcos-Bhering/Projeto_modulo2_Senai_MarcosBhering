A pós indentificar todos os produtos unicos crie uma separção para poder inserir os resultados na tabela Generico aonde diversos produtos não foram classificados

```python
palavras_genericos = [
    'CLORIDRATO', 'SÓDICO', 'MALEATO', 'BESILATO', 'ACETATO', 'FUMARATO',
    'NITRATO', 'BROMETO', 'TARTARATO', 'OLAMINA', 'CÁLCIO', 'MAGNÉSIO',
    'ÁCIDO', 'SULFATO', 'SÓDIO', 'POTÁSSICO', 'CLORETO'
]
```

```python
def eh_generico(nome):
    nome = nome.upper()
    
    # se contém palavras típicas de princípio ativo → genérico
    for p in palavras_genericos:
        if p in nome:
            return 'Sim'
    
    # se contém palavras que indicam material/equipamento → não é medicamento
    materiais = ['SERINGA', 'CATETER', 'LUVA', 'TUBO', 'APARELHO', 'MÁSCARA',
                 'EQUIPO', 'FRASCO', 'BROCA', 'AGULHA', 'CURATIVO', 'VENTILADOR',
                 'CÂNULA', 'DRENO', 'BOLSA', 'COMPRESSA', 'ODONTOLÓGICO']
    
    for m in materiais:
        if m in nome:
            return 'Não é medicamento'
    
    # se não caiu em nenhuma regra, assume não genérico
    return 'Não'

df['generico'] = df['nome_produto'].apply(eh_generico) 
# apply() aplica a função eh_generico() em cada elemento da coluna 'nome_produto' e cria uma nova coluna 'generico' com os resultados.
```

Tratar (unidade_medida) pela coluna (unidade_fornecimento_capacidade), colocar uma logica (comprimido = CP, capsula = CAP, Ampola = ML, Unidade = 1 )


```python
print(list(df['unidade_medida'].unique()))
# calculo da quantidade de NAN que tem na unidade de medida, para saber se é necessário tratar ou não.
df['unidade_medida'].isnull().sum()

```

```python
df[['unidade_medida', 'unidade_fornecimento_capacidade']].head()
```

```python
df['unidade_fornecimento_capacidade'].unique()
print(list(df['unidade_fornecimento_capacidade'].unique()))
```

```python
df['unidade_medida'] = df['unidade_fornecimento_capacidade'].fillna('DESCONHECIDO')
df['unidade_medida'] = df['unidade_fornecimento_capacidade'].str.upper()

df['unidade_medida'] = df['unidade_fornecimento_capacidade'].replace({
    r'.*COMPRIMIDO.*': 'CP',
    r'.*CÁPSULA.*': 'CAP',
    
    r'.*AMPOLA.*': 'ML',
    r'.*FRASCO.*': 'ML',
    r'.*GOTA.*': 'ML',
    r'.*SPRAY.*': 'ML',
    r'.*XAROPE.*': 'ML',
    r'.*BOLSA.*': 'ML',
    r'.*TUBETE.*': 'ML',

    r'.*POMADA.*': 'G',
    r'.*CREME.*': 'G',
    r'.*BISNAGA.*': 'G',
    r'.*POTE.*': 'G',

    r'.*PASTILHA.*': 'UNI',
    r'.*DRÁGEA.*': 'UN',
    r'.*SACHÊ.*': 'UN',
    r'.*SERINGA.*': 'UN',
    r'.*CAIXA.*': 'UN',
    r'.*EMBALAGEM.*': 'UN'
}, regex=True)
# esse (r) antes da string indica que é uma raw string, ou seja, 
# uma string literal onde os caracteres de escape (como \n, \t, etc.) não são interpretados. 
# Isso é útil quando se trabalha com expressões regulares, pois evita que o Python interprete os caracteres de escape.
```

```python
df.rename(columns={'unidade_fornecimento': 'tipo_produto'}, inplace=True) 
# essa coluna pode ajudar no power bi a fazer uma filtragem de produtos que não são medicamentos e ex: conte ampolas de forma diferentes pois tem varios tipos de ml , 
# como por exemplo: seringas, cateteres, luvas, tubos, aparelhos, máscaras, equipos, frascos, brocas, agulhas, curativos, ventiladores, cânulas, drenos, bolsas, compressas e odontológicos.
```

```python
df['unidade_medida'] = df['unidade_medida'].replace({
    r'.*UNIDADE.*': 'UN'
}, regex=True)
# sem o regex=True, o pandas iria procurar exatamente a string '.*UNIDADE.*' e não encontraria nada,
# mas com regex=True, ele interpreta a string como uma expressão regular, onde '.*' significa "qualquer coisa antes e depois de UNIDADE", 
# então ele vai substituir qualquer valor que contenha a palavra 'UNIDADE' por 'UN'.
```
